```python
# Initialize SparkSession
spark = pyspark.sql.SparkSession.builder \
    .appName("dev") \
    .master("local[*]") \
    .getOrCreate()

# Set log level to ERROR to hide warnings
spark.sparkContext.setLogLevel("ERROR")
```

```python
fin = spark.read.parquet(*glob.glob("datamart/silver/financials/*"))
inc = fin.select("Annual_Income").dropna()

inc.select(
    F.skewness("Annual_Income").alias("skew_raw"),
    F.skewness(F.log1p("Annual_Income")).alias("skew_log1p"),
).show()
print(inc.approxQuantile("Annual_Income", [0.5, 0.95, 0.99, 1.0], 0.01))

pdf = inc.toPandas()
fig, ax = plt.subplots(1, 2, figsize=(10, 3))
ax[0].hist(pdf.Annual_Income, bins=50); ax[0].set_title("raw")
ax[1].hist(np.log1p(pdf.Annual_Income), bins=50); ax[1].set_title("log1p")
plt.show()
```
